### 1. Introduction

When discussing quantum computers in the mainstream, it is often easy to get caught up in the so-called _"quantum hype"_. Apparently quantum computers will cure cancer, re-invent the internet and solve the problem of human caused climate change. This is not true. Quantum computers are an experimental technology with promising applications in specialised fields that involve very specific computational problems. Another myth, believed by many who consume popular science media, is that quantum computers offer a speed up in comparision to classical computers; we call this **quantum advanage**. As of writing, there has not yet been a quantum computer which exhibits quantum advantage on problems that are considered commercially or scientifically valuable. Quantum computers have only beaten their classical counterparts on carefully chosen computational tasks. The _"golden child"_ of quantum computing is the simulation of physical phenomena in the quantum physics and chemistry. It is well documented that classical computers begin to stuggle when simulating quantum phenomena. In particular, many body quantum simulations require [...]. Classical computers cannot intrinsically mimick quantum interactions. On the other hand, quantum computers operate on quantum principles and therefore have quantum physics _baked-in_. Theoretically, this makes them more efficient at simulating quantum processes. Another closely related application is that of optimisation. Future quantum algorithms may find the ground state energy of a system described a Hamiltonian more efficiently in comparision to their classical counterparts.
