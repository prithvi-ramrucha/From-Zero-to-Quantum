### 1. Introduction

When discussing quantum computers in the mainstream, it is often easy to get caught up in the so-called _"quantum hype"_. Apparently quantum computers will cure cancer, re-invent the internet and solve the problem of human caused climate change. This is not true. Quantum computers are an experimental technology with promising applications in specialised fields that involve very specific computational problems. Another myth, believed by many who consume popular science media, is that quantum computers offer a speed up in comparision to classical computers; often called **quantum advanage**. As of writing, there has not yet been a quantum computer which exhibits quantum advantage on problems that are considered commercially or scientifically valuable. Quantum computers have only beaten their classical counterparts on carefully chosen computational tasks. The _"golden child"_ of quantum computing is the simulation of physical phenomena in the quantum physics and chemistry. ### 2. Computational Power Scaling

The number of qubits in a computer computer is related to its computational power, but is not synonmous with it. There are various metrics that are used to quantify the performance of quantum computers. It is important to point out that the scale of a quantum computer is measured by the number of _programmable_ quibts available to execute quantum circuits. This is different to the total number of qubits the computer may possess. There is also qubit quality and speed to consider. Speed refers to the maximum circuits per second, also known as circuit throughput. In other words, it is how fast a QPU can execute circuits. Determining qubit quality is more involved. It is measured by qubit operations (QuOps) which tell us how many of the most computationally demanding quantum operations a system can reliably execute before accumulated errors caused by noise prevent the computation. Gate fidelities, coherence and dephasing times ($T_1$ and $T_2$) and computer architecture are important indicators of qubit quality. ### 4. Applications

As mentioned previously, the most promising application of quantum computers involves the **simulation of physical phenomena** that is quantum in nature. More specifically, quantum computers aim to calculate the ground state energy (lowest possible energy configuration) using the Hamiltonian $\hat{H}$ of the system. The evolution of the system over time, quantum dynamics, may also be computed. Classical computers stuggle simulating quantum systems as the quantum state space grows exponentially with the number of particles. Conversely, quantum computers theoretically may have algorithms that do not scale exponentially in time complexity. These algorithms could be applied to,

- Quantum chemistry
- Materials science
- Condensed matter physics
- High energy particle and nuclear physics

Another area of application may be in **optimisation problems**. Many problems are NP-hard in complexity theory where classical computers struggle to produce solutions for large instances $N$. It is an open question whether there exists quantum algorithms that can outperform their classical counterparts on certain NP-hard problems. While is it not yet clear which optimisation problems may offer quantum advantage, this potential application area remains of major interest due to its ubiquity and the natural mapping between optimisation objectives and quantum Hamiltonians. It is a possibility that many NP  problems in classical computing are P in when solved using hypothetical quantum algorithms. Examples of areas where optimisation problems can become computationally intractable are:

- Logistic and supply chain calculations, traffic flow optimisation, airline scheduling.
- Portfolio optimisation and risk management in finance.
- Combinatorial problems.

**Solving certain classes of PDEs** is another possible application of quantum computers. PDEs such as the Navier–Stokes equations for fluid flow, Schrödinger equation for wave quantum mechanics, Maxwell's equations for electromagnetism and the Black–Scholes for financial stocks. In theory, certain quantum approaches can are believed to offer exponential or polynomial speedups under specific assumptions. However, in practice, this application of quantum computers remains more theoretical than the previous mentioned. Even more theoretical yet, is **quantum machine learning** (QMC). The idea of QMC is to explore which sections of the machine learning (ML) pipeline can be accelerated in future using quantum computers. There are two approaches: developing new machine learning models that are inherently quantum and proposing quantum alternatives in existing pipelines. Both remain difficult. The former is formidable task due to the immaturity of quantum computing while the latter is daunting because of the efficiency of existing classical ML methods.
